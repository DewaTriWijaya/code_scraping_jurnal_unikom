```python
import pandas as pd

dfCLOC_app = pd.read_csv('../cloc/cloc_app_result.csv')
dfCLOC_core = pd.read_csv('../cloc/cloc_core_result.csv')
dfCLOC_app.head()
```

```python
# ubah nama kolom 'Nama File' menjadi 'Nama Kelas' di kedua dataframe
dfCLOC_app.rename(columns={'file_name': 'Nama Kelas'}, inplace=True)
dfCLOC_core.rename(columns={'file_name': 'Nama Kelas'}, inplace=True)

# hapus kolom file_path di kedua dataframe
dfCLOC_app.drop(columns=['file_path'], inplace=True)
dfCLOC_core.drop(columns=['file_path'], inplace=True)

# ubah nama kolom 'cloc' menjadi 'CLOC' di kedua dataframe
dfCLOC_app.rename(columns={'cloc': 'CLOC'}, inplace=True)
dfCLOC_core.rename(columns={'cloc': 'CLOC'}, inplace=True)
```

```python
# gabung semua dataframe dfCLOC_app dan dfCLOC_core menjadi dfCLOC
dfCLOC = pd.concat([dfCLOC_app, dfCLOC_core], ignore_index=True)
```
