```python
import math

def calculate_mi(cc, hl, loc, cloc):
    """
    cc   : Cyclomatic Complexity
    hl   : Halstead Volume
    loc  : Lines of Code
    cloc : Comment Lines of Code
    """

    # hindari error log(0)
    if loc == 0 or hl == 0:
        return 0

    # percent comment
    per_cm = (cloc / loc) * 100

    mi = (
        171
        - 5.2 * math.log(hl)
        - 0.23 * cc
        - 16.2 * math.log(loc)
        + 50 * math.sin(math.sqrt(2.46 * per_cm))
    )

    return round(mi, 2)
```

```python
cc = 12
hl = 350.7
loc = 220
cloc = 40

mi = calculate_mi(cc, hl, loc, cloc)
print("Maintainability Index:", mi)
```

```python
# gabung semua dataframe berdasarkan 'Nama Kelas' dari data frame dfCC, dfHL, dfLOC, dfCLOC
df_merged = dfCC.merge(dfHL, on='Nama Kelas').merge(dfLOC, on='Nama Kelas').merge(dfCLOC, on='Nama Kelas')

df_merged.head()
```

```python
# calculate MI for each row and store it in a new column 'MI'
df_merged['MI'] = df_merged.apply(lambda row: calculate_mi(row['V(g)'], row['HL(V)'], row['LOC'], row['CLOC']), axis=1)
```

```python
# buat kolom baru 'Maintainability Level' berdasarkan nilai MI
df_merged['Maintainability Level'] = df_merged['MI'].apply(classify_mi)
```
