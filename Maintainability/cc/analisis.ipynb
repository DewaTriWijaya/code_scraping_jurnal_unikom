```python
import pandas as pd

dfApp = pd.read_csv('cc_per_file.csv')
dfApp.head()
```

```python
# ubah nama function_count menjadi V(g)
dfApp = dfApp.rename(columns={'cyclomatic_complexity': 'V(g)'})

# ubah nama file_name menjadi Nama Kelas
dfApp = dfApp.rename(columns={'file_name': 'Nama Kelas'})

# hapus kolom file_path
dfApp = dfApp.drop(columns=['file_path'])
```

1 - 10
Risiko Minimal

21 - 50
Risiko Tinggi
> 50
Risiko Sangat tinggi

```python
# hapus nama kelas yang tidak ada 'ExampleInstrumentedTest.kt', 'ExampleUnitTest.kt', 'FormatUtilsTest.kt', 'DateUtilsTest.kt'
dfApp = dfApp[~dfApp['Nama Kelas'].isin(['ExampleInstrumentedTest.kt', 'ExampleUnitTest.kt', 'FormatUtilsTest.kt', 'DateUtilsTest.kt'])]
dfApp.head()
```

```python
# import to excel
dfApp.to_excel('cc_result_core_final.xlsx', index=False)
```
