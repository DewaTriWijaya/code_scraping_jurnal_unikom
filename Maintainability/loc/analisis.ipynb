```python
import pandas as pd

df = pd.read_csv('loc_mi.csv')
df.head()
```

```python
# ubah nama file_name menjadi Nama Kelas
df = df.rename(columns={'file': 'Nama Kelas'})

# ubah nama kolom logical_loc menjadi LOC
df = df.rename(columns={'loc_mi': 'LOC'})
```

```python
df.head()
```

```python
# hapus nama kelas yang tidak ada 'ExampleInstrumentedTest.kt', 'ExampleUnitTest.kt', 'FormatUtilsTest.kt', 'DateUtilsTest.kt'
df = df[~df['Nama Kelas'].isin(['ExampleInstrumentedTest.kt', 'ExampleUnitTest.kt', 'FormatUtilsTest.kt', 'DateUtilsTest.kt'])]
df.head()
```

```python
# export ke excel
df.to_excel('loc_core_result_analysis.xlsx', index=False)
```
