V = N × log2(n)

```python
import pandas as pd

df = pd.read_csv('halstead_defensible.csv')
df.head()
```

```python
# ubah nama kolom 'file_name' menjadi 'Nama Kelas'
df = df.rename(columns={'file': 'Nama Kelas'})

# ubah nama kolom 'halstead_volume' menjadi 'Halstead Volume'
df = df.rename(columns={'volume': 'V'})

# kolom menjadi dua hanya 'Nama Kelas' dan 'Halstead Volume'
df = df[['Nama Kelas', 'V']]
```

```python
df.head()
```

```python
# hapus nama kelas yang tidak ada 'ExampleInstrumentedTest.kt', 'ExampleUnitTest.kt', 'FormatUtilsTest.kt', 'DateUtilsTest.kt'
df = df[~df['Nama Kelas'].isin(['ExampleInstrumentedTest.kt', 'ExampleUnitTest.kt', 'FormatUtilsTest.kt', 'DateUtilsTest.kt'])]
df.head()
```

```python
# import menjadi export ke excel
df.to_excel('halstead_core_result_analysis.xlsx', index=False)
```
